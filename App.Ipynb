create a function called favorite_country that takes an input from the user, using the input gotten from the user compare the user input with countries_of_choice, check if the user input is available in the countries_of_choice, if there, print available else print not available.

```python
countries_of_choice = []
countries_of_choice.append("Nigeria")
countries_of_choice.append("Ghana")
countries_of_choice.append("Canada")
countries_of_choice.append("United Kingdom")
countries_of_choice.append("United States")

def favorite_country():
    country = input("Enter your favorite country: ")
    if country in countries_of_choice:
        print("Available")
    else:
        print("Not available")

favorite_country()
```
